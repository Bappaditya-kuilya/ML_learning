```python
# 1. Write a Python program that takes two numbers as input and prints their sum, difference, product, quotient, floor quotient, remainder, and power.

a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))

print("Sum: ", a+b)
print("Difference: ", a-b)
print("Product: ", a*b)
print("Quotient: ", a/b)
print("Floor Quotient: ", a//b)
print("Remainder: ", a%b)
print("Power: ", a**b)
```

```python
# 2.Write a Python program that stores your name, age, height, and whether you are a student. Print each value along with its data type.

name = input("Name: ")
age = int(input("Age: "))
height = float(input("Height: "))
is_student_input = input("Are you a student? (yes/no): ")
is_student = is_student_input.lower() == 'yes'

print("Name:", name)
print("Type:", type(name))
print("Age:", age)
print("Type:", type(age))
print("Height:", height)
print("Type:", type(height))
print("Is Student:", is_student)
print("Type:", type(is_student))
```

```python
# 3.Write a Python program that takes a three-digit number and prints the sum of its digits.

num = int(input("Enter a 3 digit number: "))
sum=0
rem=0

for i in range(4):
  rem = num%10
  num = num // 10
  sum = sum + rem
print(sum)
```

```python
# 4.Write a Python program that takes the length and width of a rectangle and prints its area, perimeter, and diagonal length.

import math

length = float(input("Length: "))
width = float(input("Width: "))

area = length * width
print("Area: ",area)

perimeter = 2*(length+width)
print("Perimeter: ",perimeter)

dia_length = math.sqrt(length**2 + width**2)
print("Diagonal Length: ",dia_length)

```

```python
# 5. Write a Python program that takes two integers and swaps their values without using a third variable.

a = int(input("Enter First No:"))
b = int(input("Enter Second No:"))

print("Before :", a ,b)
a,b = b,a

print("After :", a,b)
```
